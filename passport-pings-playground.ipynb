```python
# Passport Pings — notebook starter
# Requires Python 3.9 or newer.

# A country, city, and airport are featured destinations.
# They do not imply that a language belongs only to that location.
greetings = [
    {
        "greeting": "Namaste",
        "language": "Hindi",
        "meaning": "Hello / a respectful greeting",
        "country_code": "IN",
        "city": "Mumbai",
        "airport_code": "BOM",
    },
    {
        "greeting": "Konnichiwa",
        "language": "Japanese",
        "meaning": "Hello / good afternoon",
        "country_code": "JP",
        "city": "Tokyo",
        "airport_code": "NRT",
    },
    {
        "greeting": "Ciao",
        "language": "Italian",
        "meaning": "Hello / goodbye",
        "country_code": "IT",
        "city": "Rome",
        "airport_code": "FCO",
    },
]


def country_flag(country_code: str) -> str:
    """Convert a two-letter country code into a flag emoji."""

    # Check the format, not whether the country code actually exists.
    if len(country_code) != 2 or any(
        letter not in "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        for letter in country_code
    ):
        raise ValueError(
            "Use a two-letter uppercase country code, such as IN."
        )

    # Flag emojis are pairs of Unicode regional-indicator characters.
    # ord() converts a character to a number; chr() does the reverse.
    return "".join(
        chr(0x1F1E6 + ord(letter) - ord("A"))
        for letter in country_code
    )


def format_ping(entry: dict[str, str]) -> str:
    """Format one greeting as a Passport Pings message."""

    flag = country_flag(entry["country_code"])

    # f-strings insert values into text.
    # \n creates a line break.
    return (
        f"{flag} PASSPORT PINGS\n"
        f"{entry['greeting']}!\n\n"
        f"Language: {entry['language']}\n"
        f"Meaning: {entry['meaning']}\n"
        f"Destination: {entry['city']} ({entry['airport_code']})\n"
        f"Featured country: {entry['country_code']}"
    )

# Main caller function - takes inpoout list if dic where key,val are strings and index integer
# it returns tuple
def generate_ping(
    greetings: list[dict[str, str]],
    next_index: int,
) -> tuple[str, int]:
    """Return the current message and the next selection position."""

    if not greetings:
        raise ValueError("Add at least one greeting.")

    if not 0 <= next_index < len(greetings):
        raise ValueError(
            "The selection position is outside the greeting list."
        )

    selected_greeting = greetings[next_index]
    message = format_ping(selected_greeting)

    # Modulo (%) wraps back to zero after the final greeting.
    updated_index = (next_index + 1) % len(greetings)

    # Return state explicitly instead of changing a global variable.
    return message, updated_index


# Start at the first greeting. Python list positions begin at zero.
next_index = 0

# Call the function and unpack its two returned values.
message, next_index = generate_ping(greetings, next_index)

print(message)
```

```python
# Passport Pings — full-year notebook starter
# Requires Python 3.9+; no extra packages or API keys.
#
# This prints messages only.
# It does not save progress, create commits, or notify Discord.
#
# Language data is a draft: review it before automated publication.
# Latin-script spelling retains accents. Greek uses romanization.


# Each row contains:
# language, featured country, featured city, IATA airport code,
# and five phrases.
#
# Phrase positions generally represent:
# hello, morning, evening, good night, goodbye.
# Exceptions are handled in meaning_overrides below.
destination_data = [
    (
        "English", "US", "San Francisco", "SFO",
        ("Hello", "Good morning", "Good evening",
         "Good night", "Goodbye"),
    ),
    (
        "French", "FR", "Paris", "CDG",
        ("Salut", "Bonjour", "Bonsoir",
         "Bonne nuit", "Au revoir"),
    ),
    (
        "Spanish", "ES", "Madrid", "MAD",
        ("Hola", "Buenos días", "Buenas tardes",
         "Buenas noches", "Adiós"),
    ),
    (
        "German", "DE", "Frankfurt", "FRA",
        ("Hallo", "Guten Morgen", "Guten Abend",
         "Gute Nacht", "Auf Wiedersehen"),
    ),
    (
        "Italian", "IT", "Rome", "FCO",
        ("Ciao", "Buongiorno", "Buonasera",
         "Buonanotte", "Arrivederci"),
    ),
    (
        "Portuguese", "PT", "Lisbon", "LIS",
        ("Olá", "Bom dia", "Boa tarde",
         "Boa noite", "Adeus"),
    ),
    (
        "Dutch", "NL", "Amsterdam", "AMS",
        ("Hallo", "Goedemorgen", "Goedenavond",
         "Welterusten", "Tot ziens"),
    ),
    (
        "Swedish", "SE", "Stockholm", "ARN",
        ("Hej", "God morgon", "God kväll",
         "God natt", "Hej då"),
    ),
    (
        "Norwegian Bokmål", "NO", "Oslo", "OSL",
        ("Hei", "God morgen", "God kveld",
         "God natt", "Ha det"),
    ),
    (
        "Danish", "DK", "Copenhagen", "CPH",
        ("Hej", "Godmorgen", "Godaften",
         "Godnat", "Farvel"),
    ),
    (
        "Finnish", "FI", "Helsinki", "HEL",
        ("Hei", "Hyvää huomenta", "Hyvää iltaa",
         "Hyvää yötä", "Näkemiin"),
    ),
    (
        "Icelandic", "IS", "Reykjavik", "KEF",
        ("Halló", "Góðan morgun", "Gott kvöld",
         "Góða nótt", "Bless"),
    ),
    (
        "Polish", "PL", "Warsaw", "WAW",
        ("Cześć", "Dzień dobry", "Dobry wieczór",
         "Dobranoc", "Do widzenia"),
    ),
    (
        "Czech", "CZ", "Prague", "PRG",
        ("Ahoj", "Dobré ráno", "Dobrý večer",
         "Dobrou noc", "Na shledanou"),
    ),
    (
        "Slovak", "SK", "Bratislava", "BTS",
        ("Ahoj", "Dobré ráno", "Dobrý večer",
         "Dobrú noc", "Dovidenia"),
    ),
    (
        "Slovenian", "SI", "Ljubljana", "LJU",
        ("Živjo", "Dobro jutro", "Dober večer",
         "Lahko noč", "Nasvidenje"),
    ),
    (
        "Croatian", "HR", "Zagreb", "ZAG",
        ("Bok", "Dobro jutro", "Dobra večer",
         "Laku noć", "Doviđenja"),
    ),
    (
        "Serbian", "RS", "Belgrade", "BEG",
        ("Zdravo", "Dobro jutro", "Dobro veče",
         "Laku noć", "Doviđenja"),
    ),
    (
        "Albanian", "AL", "Tirana", "TIA",
        ("Përshëndetje", "Mirëmëngjes", "Mirëmbrëma",
         "Natën e mirë", "Mirupafshim"),
    ),
    (
        "Romanian", "RO", "Bucharest", "OTP",
        ("Salut", "Bună dimineața", "Bună seara",
         "Noapte bună", "La revedere"),
    ),
    (
        "Hungarian", "HU", "Budapest", "BUD",
        ("Szia", "Jó reggelt", "Jó estét",
         "Jó éjszakát", "Viszontlátásra"),
    ),
    (
        "Greek (romanized)", "GR", "Athens", "ATH",
        ("Geia sou", "Kalimera", "Kalispera",
         "Kalinychta", "Antio"),
    ),
    (
        "Turkish", "TR", "Istanbul", "IST",
        ("Merhaba", "Günaydın", "İyi akşamlar",
         "İyi geceler", "Hoşça kal"),
    ),
    (
        "Estonian", "EE", "Tallinn", "TLL",
        ("Tere", "Tere hommikust", "Tere õhtust",
         "Head ööd", "Head aega"),
    ),
    (
        "Latvian", "LV", "Riga", "RIX",
        ("Sveiki", "Labrīt", "Labvakar",
         "Ar labu nakti", "Uz redzēšanos"),
    ),
    (
        "Lithuanian", "LT", "Vilnius", "VNO",
        ("Labas", "Labas rytas", "Labas vakaras",
         "Labanakt", "Viso gero"),
    ),
    (
        "Irish", "IE", "Dublin", "DUB",
        ("Dia dhuit", "Maidin mhaith", "Tráthnóna maith",
         "Oíche mhaith", "Slán"),
    ),
    (
        "Welsh", "GB", "Cardiff", "CWL",
        ("Helo", "Bore da", "Noswaith dda",
         "Nos da", "Hwyl fawr"),
    ),
    (
        "Indonesian", "ID", "Jakarta", "CGK",
        ("Halo", "Selamat pagi", "Selamat sore",
         "Selamat tidur", "Sampai jumpa"),
    ),
    (
        "Malay", "MY", "Kuala Lumpur", "KUL",
        ("Hai", "Selamat pagi", "Selamat petang",
         "Selamat tidur", "Jumpa lagi"),
    ),
    (
        "Vietnamese", "VN", "Hanoi", "HAN",
        ("Xin chào", "Chào buổi sáng", "Chào buổi tối",
         "Chúc ngủ ngon", "Tạm biệt"),
    ),
    (
        "Swahili", "KE", "Nairobi", "NBO",
        ("Habari", "Habari za asubuhi", "Habari za jioni",
         "Usiku mwema", "Kwaheri"),
    ),
]


# Default meanings for the five positions.
default_meanings = (
    "Hello",
    "Good morning",
    "Good evening",
    "Good night",
    "Goodbye",
)


# Keep these notes general: politeness and usage vary by language.
default_notes = (
    "A greeting; formality depends on the language and situation.",
    "A morning greeting or wish.",
    "An evening greeting or wish.",
    "A nighttime or bedtime wish, often used when taking leave.",
    "A farewell; politeness depends on the situation.",
)


# A tuple key identifies a particular language and phrase position.
# Positions start at zero.
#
# These exceptions prevent us from assigning identical English meanings
# to phrases whose meanings differ.
meaning_overrides = {
    ("French", 0): (
        "Hi / bye",
        "Informal; suitable for familiar people.",
    ),
    ("French", 1): (
        "Hello / good day",
        "A common daytime greeting, including in the morning.",
    ),
    ("Spanish", 2): (
        "Good afternoon",
        "Afternoon usage varies by region.",
    ),
    ("Spanish", 3): (
        "Good evening / good night",
        "Can be used when greeting or taking leave at night.",
    ),
    ("Italian", 0): (
        "Hello / goodbye",
        "Informal; suitable for familiar people.",
    ),
    ("Italian", 1): (
        "Good morning / good day",
        "A polite daytime greeting.",
    ),
    ("Portuguese", 2): (
        "Good afternoon",
        "An afternoon greeting.",
    ),
    ("Portuguese", 3): (
        "Good evening / good night",
        "Can be used when greeting or taking leave at night.",
    ),
    ("Dutch", 3): (
        "Sleep well / good night",
        "A bedtime wish.",
    ),
    ("Polish", 0): (
        "Hi / bye",
        "An informal greeting or farewell.",
    ),
    ("Polish", 1): (
        "Good morning / good day",
        "A polite daytime greeting.",
    ),
    ("Hungarian", 0): (
        "Hi / bye",
        "Informal, addressed to one person.",
    ),
    ("Greek (romanized)", 0): (
        "Hello / goodbye",
        "Informal singular; romanization conventions vary.",
    ),
    ("Turkish", 4): (
        "Goodbye / stay well",
        "Said by the person leaving to someone staying.",
    ),
    ("Irish", 2): (
        "Good afternoon / good evening",
        "An afternoon or evening greeting.",
    ),
    ("Indonesian", 2): (
        "Good late afternoon",
        "Used in the late afternoon.",
    ),
    ("Indonesian", 3): (
        "Sleep well / good night",
        "Said to someone going to sleep.",
    ),
    ("Indonesian", 4): (
        "See you",
        "A farewell.",
    ),
    ("Malay", 2): (
        "Good afternoon / good evening",
        "Used during the afternoon or early evening.",
    ),
    ("Malay", 3): (
        "Sleep well / good night",
        "Said to someone going to sleep.",
    ),
    ("Malay", 4): (
        "See you again",
        "A farewell.",
    ),
    ("Swahili", 0): (
        "How are things? / what's the news?",
        "A conversational greeting inviting a response.",
    ),
    ("Swahili", 1): (
        "How is your morning?",
        "A morning inquiry used as a greeting.",
    ),
    ("Swahili", 2): (
        "How is your evening?",
        "An evening inquiry used as a greeting.",
    ),
}


def build_greetings() -> list[dict[str, str]]:
    """Expand the compact dataset into our familiar greeting dictionaries."""

    collection: list[dict[str, str]] = []

    # Put the first phrase from every destination first, then the second,
    # and so on. This avoids five consecutive pings from the same place.
    for phrase_index in range(len(default_meanings)):
        for language, country, city, airport, phrases in destination_data:
            if len(phrases) != len(default_meanings):
                raise ValueError(
                    f"{language}: expected five phrases."
                )

            meaning, note = meaning_overrides.get(
                (language, phrase_index),
                (
                    default_meanings[phrase_index],
                    default_notes[phrase_index],
                ),
            )

            collection.append(
                {
                    "greeting": phrases[phrase_index],
                    "language": language,
                    "meaning": meaning,
                    "usage_note": note,
                    "country_code": country,
                    "city": city,
                    "airport_code": airport,
                }
            )

    return collection


def country_flag(country_code: str) -> str:
    """Convert an uppercase two-letter country code into a flag emoji."""

    if len(country_code) != 2 or any(
        letter not in "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
        for letter in country_code
    ):
        raise ValueError(
            "Use a two-letter uppercase country code, such as IN."
        )

    # This checks code format, not membership in a country-code registry.
    return "".join(
        chr(0x1F1E6 + ord(letter) - ord("A"))
        for letter in country_code
    )


def validate_greetings(
    collection: list[dict[str, str]],
) -> None:
    """Check fields, code formats, and duplicate language/phrase entries."""

    if not collection:
        raise ValueError("The greeting collection is empty.")

    required_fields = (
        "greeting",
        "language",
        "meaning",
        "usage_note",
        "country_code",
        "city",
        "airport_code",
    )

    seen: set[tuple[str, str]] = set()

    for index, entry in enumerate(collection):
        for field in required_fields:
            value = entry.get(field)

            if not isinstance(value, str) or not value.strip():
                raise ValueError(
                    f"Entry {index}: '{field}' must contain text."
                )

        country_flag(entry["country_code"])

        airport = entry["airport_code"]

        if len(airport) != 3 or any(
            letter not in "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
            for letter in airport
        ):
            raise ValueError(
                f"Entry {index}: airport code must be three uppercase letters."
            )

        # The same phrase in the same language is a duplicate,
        # even if someone assigns it a different city.
        key = (
            entry["language"].strip().casefold(),
            entry["greeting"].strip().casefold(),
        )

        if key in seen:
            raise ValueError(
                f"Duplicate: {entry['language']} / {entry['greeting']}"
            )

        seen.add(key)


def format_ping(
    entry: dict[str, str],
    ping_number: int,
) -> str:
    """Build a Passport Pings message without printing or sending it."""

    flag = country_flag(entry["country_code"])

    return (
        f"{flag} PASSPORT PINGS — #{ping_number}\n"
        f"{entry['greeting']}!\n\n"
        f"Language: {entry['language']}\n"
        f"Meaning: {entry['meaning']}\n"
        f"Usage: {entry['usage_note']}\n"
        f"Featured destination: {entry['city']} "
        f"({entry['airport_code']})\n"
        f"Featured country: {entry['country_code']}"
    )


def generate_ping(
    collection: list[dict[str, str]],
    next_turn: int,
) -> tuple[str, int]:
    """Return the next message and updated turn counter."""

    if not collection:
        raise ValueError("Add at least one greeting.")

    if next_turn < 0:
        raise ValueError("The turn counter cannot be negative.")

    # Modulo wraps back to the first entry after the collection is exhausted.
    entry_index = next_turn % len(collection)
    entry = collection[entry_index]

    message = format_ping(
        entry,
        ping_number=next_turn + 1,
    )

    return message, next_turn + 1


# Build and structurally validate the dataset.
# Validation cannot establish translation accuracy.
greetings = build_greetings()
validate_greetings(greetings)

# Explicitly check that this starter meets our full-year capacity target.
if len(greetings) < 160:
    raise ValueError("At least 160 entries are required for this starter.")

print(f"Loaded {len(greetings)} language/phrase entries.")
print(f"At three pings per week: {len(greetings) / 3:.1f} weeks.")
print()

# Run once to initialize. Rerunning the whole cell resets progress.
next_turn = 0

message, next_turn = generate_ping(greetings, next_turn)
print(message)
```
